### Step 1: Load and Standardize the Data
Before applying PCA, we must standardize the dataset. Standardization ensures that all features have a mean of 0 and a standard deviation of 1, which is essential for PCA.
Fill in the code to standardize the dataset.

STRICTLY - Write code that implements standardization based on the image below




```python
import numpy as np
import pandas as pd

data = pd.read_csv("african_pca_dataset.csv")

```

```python
X = data.drop("Country", axis=1)
X = X.fillna(X.mean())

```

```python
mean = np.mean(X, axis=0)
std = np.std(X, axis=0, ddof=1)
```

```python
standardized_data = (X - mean) / std
standardized_data[:5]
```

### Step 3: Calculate the Covariance Matrix
The covariance matrix helps us understand how the features are related to each other. It is a key component in PCA.

```python
cov_matrix = np.cov(standardized_data, rowvar=False)
cov_matrix
```

In a paragraph that has a maximum if 5 Lines, Explain why we need to compute a covariance matrix, provide atleast 2 reasons

### Step 4: Perform Eigendecomposition
Eigendecomposition of the covariance matrix will give us the eigenvalues and eigenvectors, which are essential for PCA.
Fill in the code to compute the eigenvalues and eigenvectors of the covariance matrix.

```python
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)
eigenvalues, eigenvectors
```

### Step 5: Sort Principal Components
Sort the eigenvectors based on their corresponding eigenvalues in descending order. The higher the eigenvalue, the more important the eigenvector.
Complete the code to sort the eigenvectors and print the sorted components.

<a url ='https://www.youtube.com/watch?v=vaF-1xUEXsA&t=17s'>How Is Explained Variance Used In PCA?'<a/>

```python
sorted_indices = np.argsort(eigenvalues)[::-1]
```

```python
sorted_eigenvectors = eigenvectors[:, sorted_indices]
```

```python
sorted_eigenvectors
```

The covariance matrix is computed to understand how variables vary together and to identify relationships between features. PCA uses these relationships to find directions that capture the greatest variation in the data. It also helps reduce redundancy by detecting correlated variables. Without the covariance matrix, it would be difficult to determine the principal components that best summarize the dataset.

### Step 6: Project Data onto Principal Components
Now that we’ve selected the number of components, we will project the original data onto the chosen principal components.
Fill in the code to perform the projection.

```python
num_components = 4
principal_components = sorted_eigenvectors[:, :num_components]
```

```python
reduced_data = np.dot(standardized_data, principal_components)
reduced_data[:5]
```

### Step 7: Output the Reduced Data
Finally, display the reduced data obtained by projecting the original dataset onto the selected principal components.

```python
print(f"Reduced Data Shape: {reduced_data.shape}")
reduced_data[:5]
```

### Step 8: Visualize Before and After PCA
Now, let's plot the original data and the data after PCA to compare the reduction in dimensions visually.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.scatter(
    standardized_data.values[:,0],
    standardized_data.values[:,1]
)
plt.title("Original Data")
plt.xlabel("GDP per Capita")
plt.ylabel("Population")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,5))

plt.subplot(1,2,2)
plt.scatter(
    reduced_data[:,0],
    reduced_data[:,1]
)
plt.title("After PCA")
plt.xlabel("PC1")
plt.ylabel("PC2")
```
